## Data Cleaning


Removing Duplicate Value

```python
data = data.drop_duplicates()
```

```python
data.duplicated().sum()
```

Cleaning Column Names

```python
new = data.columns


new

```

```python
for i in range(len(new)):
   
    
    # 1. Use square brackets data[c] to use the variable
    print(data[new[i]].dtype)  
    
    # 2. Check for 'object' (how pandas stores strings)
    if data[new[i]].dtype == 'object': 
        # 3. Clean the text and reassign it back using brackets
        data[new[i]] = data[new[i]].str.strip().str.title()
      
data
```

Fix Nagative Billing Amount


```python
data.loc[data['Billing Amount']<0,"Billing Amount"] = data['Billing Amount'].mean()
```

```python
data.head()
```

```python
data['Date of Admission'] = pd.to_datetime(data['Date of Admission'])
```

```python
data['Discharge Date'] = pd.to_datetime(data['Discharge Date'])
```

```python
data['Length_of_stay'] = ( data['Discharge Date']- data['Date of Admission']).dt.days
```
