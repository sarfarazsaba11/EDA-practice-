## Data Cleaning


Removing Duplicate Value

```python
data = data.drop_duplicates()
```

```python
data.duplicated().sum()
```

Cleaning Column Names

```python
new = data.columns


new

```

```python
for i in range(len(new)):
   
    
    # 1. Use square brackets data[c] to use the variable
    print(data[new[i]].dtype)  
    
    # 2. Check for 'object' (how pandas stores strings)
    if data[new[i]].dtype == 'object': 
        # 3. Clean the text and reassign it back using brackets
        data[new[i]] = data[new[i]].str.strip().str.title()
      
data
```

Fix Nagative Billing Amount


```python
data.loc[data['Billing Amount']<0,"Billing Amount"] = data['Billing Amount'].mean()
```

```python
data.head()
```

```python
data['Date of Admission'] = pd.to_datetime(data['Date of Admission'])
```

```python
data['Discharge Date'] = pd.to_datetime(data['Discharge Date'])
```

```python
data['Length_of_stay'] = ( data['Discharge Date']- data['Date of Admission']).dt.days
```

## How has average treatment cost changed over time ?

```python
average_cost = data ['Billing Amount'].mean()

average_cost
```

```python
data['Admission_month'] = data['Date of Admission'].dt.to_period("M")
data.dtypes

data.head()
```

```python
monthly_average_cost = data.groupby("Admission_month")['Billing Amount'].mean()

monthly_average_cost
```

```python
plt.figure(figsize=(20 , 5))
plt.plot(monthly_average_cost.index.astype(str), monthly_average_cost.values,  color= 'red', marker = 'o', markersize= 3)
plt.title("Average Treatment Cost (Montly)")
plt.xlabel("Month")
plt.ylabel("Billing amount")
plt.tight_layout
plt.xticks(rotation = 90)
plt.show()
```

## Is there any relationship between patient age and treatment cost ?

```python
average_bill_by_age = data.groupby("Age")["Billing Amount"].mean()

average_bill_by_age

```

```python
plt.figure(figsize=(30, 5))
plt.plot(average_bill_by_age.index.astype(str), average_bill_by_age.values,  color= 'red', marker = 'o', markersize= 3)
plt.title("Average Treatment Cost by Age")
plt.xlabel("Age")
plt.ylabel("Billing amount")
plt.tight_layout
plt.xticks(rotation = 90)
plt.show()
```

```python
sample_data = data.sample(100, random_state =1)

plt.figure(figsize=(20, 5))

plt.scatter(sample_data['Age'], sample_data['Billing Amount'] )

plt.title("Treatment Cost and  Age Relationship")
plt.xlabel("Age")
plt.ylabel("Billing Amount")
plt.tight_layout
plt.show()
```
